# Exercise 2 - Load a CSV, display and plot
Reading models.csv with pandas, showing the table, then charting accuracy.

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("models.csv")
df
```

```python
df.plot(x="model", y="accuracy", kind="bar", legend=False)
plt.title("Model accuracy")
plt.ylabel("accuracy")
plt.show()
```
